1. What is nf-core?

- nf-core is an international open-source bioinformatics community
- it develops reusable Nextflow pipelines, modules, and subworkflows
- pipelines follow common standards and best-practice guidelines
- they are designed to be portable, scalable, tested, and reproducible
- the code is community developed and openly available

https://nf-co.re/docs/get_started/nf-core

---

- Released: 101
- Under development: 43
- Archived: 13

https://nf-co.re/pipelines/

3. Are there any non-bioinformatic pipelines in nf-core?

- yes
- **`nf-core/rangeland`** is an **Earth Observation** pipeline that processes satellite imagery to study land-cover changes
- **`nf-core/meerpipe`** is an **astronomy pipeline** that processes MeerKAT pulsar data
- **`nf-core/spinningjenny`** simulates the **Industrial Revolution using agent-based models**
- **`nf-core/datasync`** is for **system operations/data-transfer automation**, rather than biological analysis

https://nf-co.re/pipelines/

---

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

- a pipeline is a sequence of connected computational steps that processes data from input to final output.
- it combines multiple tasks to automate complex analyses
- the output of one step can become the input of the next
- pipelines make analyses more **consistent and reproducible

https://www.enthought.com/blog/demystifying-bioinformatics-pipelines

---

5. Why do you think nf-core adheres to strict guidelines?

- strict guidelines keep nf-core pipelines consistent, even when developed by different people
- they improve reproducibility and reliability through standardized practices and testing
- they make pipelines easier to understand, maintain, and reuse
- they help pipelines run consistently across different computing environments

---

6. What are the main features of nf-core pipelines?

- **Standardized:** pipelines follow common nf-core guidelines and structures.
- **Reproducible:** software versions, parameters, and workflow steps are controlled.
- **Portable and scalable:** they can run on laptops, HPC clusters, and cloud systems.
- **Modular:** pipelines use reusable modules and subworkflows.
- **Tested and documented:** pipelines include automated tests and detailed documentation.
- **Containerized:** dependencies can be managed with Docker or Apptainer/Singularity.
- **Configurable and community reviewed:** users can adjust settings, and pipelines are openly reviewed by the nf-core community.

https://nf-co.re

**What changed**: "nothing"

- pipeline ran again from scratch because `-resume` was not used.
- results stayed the same because the pipeline, input, and parameters were unchanged
- the 2. run was slightly faster because Docker images were already cached

---

```python
# now set -resume to the command. What did change?
!nextflow run nf-core/differentialabundance -r 2.0.0 -profile test,docker --outdir test_results -resume
```

**What changed**: existing intermediate results in the work directory were reused.

- `-resume` reused cached results from previously completed processes
- much less computation was needed, the run finished much faster
- final results stayed the same

---

Check out the current directory. Next to the outdir you specified, what else has changed?

Next to the output directory, Nextflow created a `work` directory, a `.nextflow` directory, and log files 

- `work/`: stores **intermediate process files** and cached results used by `-resume`
- `.nextflow/`: stores **Nextflow metadata/cache information**
- `.nextflow.log`: log of the **latest run**
- `.nextflow.log.1`, `.2`, `.3`: logs from **previous runs**

---

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!nextflow run nf-core/differentialabundance -r 2.0.0 -profile test,docker --outdir test_results -resume
```

**What changed**

- deleting `work/` removed the cached intermediate results
- `-resume` therefore could not reuse previous processes
- all 25 processes ran again
- the pipeline still completed successfully
- `-resume` only works if the relevant files in `work/` still exist

### What is differential abundance analysis?

- differential abundance analysis compares how much of a feature is present between groups
- it identifies features that are more or less abundant in one group
- examples of features are genes, proteins, cell types, or microbes
- it helps find biological differences between conditions
