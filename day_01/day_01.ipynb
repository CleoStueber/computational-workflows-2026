1. What is nf-core?

- **nf-core is an international open-source community** focused on bioinformatics workflows.
- It develops **Nextflow pipelines, modules, and subworkflows** that can be reused for different analyses.
- It provides **common standards, templates, and best-practice guidelines** so workflows are built in a consistent way.
- nf-core pipelines are intended to be **portable and scalable**, so they can run on laptops, HPC systems, and cloud platforms.
- Pipelines are **tested, documented, and reviewed by the community**, which supports reliable and reproducible analyses.
- The code is **community owned**, and anyone can use it, contribute improvements, or help maintain it.

https://nf-co.re/docs/get_started/nf-core

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

- sequence of connected computational steps that process data from an input into a final output.
- a pipeline can combine tasks and is used to **automate complex analyses**, so individual steps do not have to be run manually each time.
- each step performs a specific task, and the **output of one step can become the input of the next**.
- make analyses **consistent and reproducible**, because the same steps can be applied to different datasets in the same way.

https://www.enthought.com/blog/demystifying-bioinformatics-pipelines

5. Why do you think nf-core adheres to strict guidelines?

- Strict guidelines help make nf-core pipelines **consistent**, even when they are developed by different people.
- They improve **reproducibility**, because pipelines follow standardized structures and practices.
- They support **reliability and quality control** through testing, documentation, and peer review.
- Standardization makes pipelines **easier to understand, maintain, and reuse**.
- It helps ensure that pipelines can run across different computing environments.

6. What are the main features of nf-core pipelines?

- **Standardized structure:** nf-core pipelines follow the same general layout and development guidelines.
- **Reproducible:** software versions, parameters, and workflow steps are controlled so analyses can be repeated reliably.
- **Portable:** pipelines can run on different systems, including laptops, HPC clusters, and cloud platforms.
- **Scalable:** they can handle both small and large datasets using available computing resources.
- **Modular:** pipelines are built from reusable modules and subworkflows.
- **Well documented:** each pipeline includes instructions, parameter descriptions, and usage information.
- **Tested:** automated tests are used to check that pipelines work correctly.
- **Container support:** software dependencies can be provided through technologies such as Docker or Singularity/Apptainer.
- **Community reviewed:** pipelines are developed openly and reviewed by the nf-core community.
- **Configurable:** users can adjust parameters and computing settings without changing the core pipeline code.

https://nf-co.re

- `-resume`: Nextflow reused results from previously completed processes

- final results stayed the same

Check out the current directory. Next to the outdir you specified, what else has changed?

Next to the output directory (`test_results/`), Nextflow created a `work` directory, a `.nextflow` directory, and log files 

- `work/` → intermediate files and results from individual pipeline processes (used by `-resume`)
- `.nextflow/` → local Nextflow metadata and execution information
- `.nextflow.log` → log file for the most recent Nextflow run
- `.nextflow.log.1`, `.nextflow.log.2`, `.nextflow.log.3` → logs from the previous runs

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!nextflow run nf-core/differentialabundance -r 2.0.0 -profile test,docker --outdir test_results -resume
```

**What changed**

- `-resume` could no longer reuse the previous intermediate results because they had been deleted with the `work` directory.
- The processes were **executed again instead of being loaded from the cache**.
- The pipeline completed successfully and ran all **25 processes**
- The `work` directory is required for Nextflow's resume functionality. 
`-resume` alone is not enough if the cached files have been deleted.

## Lets look at the results

### What is differential abundance analysis?

- Differential abundance analysis compares the abundance of features between **different groups or conditions**.
- For RNA-seq, the features are **genes**. The analysis asks which genes have significantly different expression between the groups.
- The result includes a **fold change** (size/direction of the difference) and statistical significance such as a **p-value / adjusted p-value**. 

Give the most important plots from the report:

### Important plots

The `nf-core/differentialabundance` test run generated several plots for exploratory analysis, quality control, and differential expression analysis.

- **PCA plot:** summarizes the major sources of variation between samples and shows whether samples with the same treatment tend to cluster together.
- **3D PCA plot:** shows sample separation using the first three principal components.
- **Sample clustering dendrogram:** groups samples according to similarity in gene expression and shows whether samples cluster according to treatment.
- **Boxplot:** compares gene-abundance distributions between samples before and after normalization and variance stabilization.
- **Density plot:** shows the overall distribution of abundance values and helps assess sample comparability.
- **MAD correlation plot:** shows similarity between samples and can help identify unusual or poorly correlated samples.
- **Volcano plot:** combines effect size and statistical significance to highlight differentially expressed genes.
- **DESeq2 dispersion plot:** shows the relationship between mean gene abundance and estimated dispersion and is used to assess the DESeq2 model.
